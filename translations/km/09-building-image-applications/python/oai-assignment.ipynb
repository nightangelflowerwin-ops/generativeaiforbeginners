# ការបង្កើតកម្មវិធីបង្កើតរូបភាព (OpenAI)

មានរឿងច្រើនជាងការបង្កើតអត្ថបទសម្រាប់ LLMs។ អ្នកអាចបង្កើតរូបភាពពីការពិពណ៌នាអត្ថបទផងដែរ។ រូបភាពជាមូដាល់មានប្រយោជន៍នៅក្នុងវិស័យវេជ្ជសាស្ត្របច្ចេកវិទ្យា (MedTech), សំនង់សំណឹក, ទេសចរណ៍, ការអភិវឌ្ឍហ្គេម, ផ្សព្វផ្សាយ និងជាច្រើនទៀត។ នៅក្នុងមេរៀននេះ យើង អនុវត្តជាមួយម៉ូដែល **GPT Image** សម័យបច្ចុប្បន្ននៅលើវេទិកា OpenAI។

## គោលបំណងសិក្សា

នៅចុងបញ្ចប់មេរៀននេះ អ្នកនឹងអាចធ្វើបាន៖

- អъясះពិចារណាថាបង្កើតរូបភាពប្រាប់មានអ្វី និងកន្លែងដែលវាដាក់ប្រយោជន៍។
- យល់ដឹងអំពីគ្រួសារ `gpt-image` និងភាពខុសគ្នារវាងវានិងម៉ូដែល DALL·E ជំនាន់ចាស់។
- បង្កើតកម្មវិធីបង្កើតរូបភាព និងកែរូបភាពបាន។

## បង្កើតរូបភាពគឺជាអ្វី?

ម៉ូដែលបង្កើតរូបភាពបង្កើតរូបពីការបញ្ចូលអត្ថបទ (prompt)។ ម៉ូដែលទំនើបដូចជា `gpt-image` រៀនការតភ្ជាប់រវាងអត្ថបទនិងរូបភាពនៅក្នុងដំណាក់កាលហ្វឹកហាត់បណ្តាល បន្ទាប់មកលាប់លើសំឡេងរំខានចៃដន្យដើម្បីបង្កើតរូបភាពដែលគឺត្រូវនឹងការបញ្ចូលរបស់អ្នក។

គ្រួសារម៉ូដែលរូបភាពពីរ ពោលគឺ:

- **`gpt-image` (OpenAI)** - ជាកំណើតបច្ចុប្បន្នដែលប្រើក្នុងមេរៀននេះ។ វាគាំទ្រការបង្កើតរូបពីអត្ថបទ និងការកែតម្រូវរូបភាព (ជួសជុលជាមួយម៉ាស)
- **Midjourney** - ម៉ូដែលបីភាគីល្បីមួយដែលមានសេវាកម្មផ្ទាល់ខ្លួន និងដំណើរការជាមួយ Discord។

> ម៉ូដែលរូបភាព OpenAI ចាស់ៗ - **DALL·E 2** និង **DALL·E 3** - គឺជាជំនាន់ចាស់។ DALL·E 3 មិនទាន់មានសម្រាប់ការដាក់បញ្ចូលថ្មីហើយ ហើយលក្ខណៈ `create_variation` មានតែនៅ DALL·E 2 ប៉ុណ្ណោះ។ សូមប្រើម៉ូដែល `gpt-image` សម្រាប់កម្មវិធីថ្មីៗ។

> **សំខាន់:** ម៉ូដែល `gpt-image` បញ្ចេញរូបភាពបង្កើតជារាង **base64** (`b64_json`) មិនមែន URL ទេ។ កូដរបស់អ្នកត្រូវបម្លែងខ្សែអារ៉េ base64 ទៅជាបៃ និងរក្សាទុកវា - មិនមាន URL រូបភាពសម្រាប់ទាញយកទេ។


## ការសាងសង់កម្មវិធីបង្កើតរូបភាពដំបូងរបស់អ្នក

តើយើងត្រូវការអ្វីខ្លះដើម្បីបង្កើតកម្មវិធីបង្កើតរូបភាព? អ្នកត្រូវការឧបករណ៍បណ្ណាល័យដូចខាងក្រោម៖

- **python-dotenv**, អ្នកត្រូវបានណែនាំយ៉ាងខ្លាំងឱ្យប្រើបណ្ណាល័យនេះដើម្បីរក្សារសម្ងាត់របស់អ្នកនៅក្នុងឯកសារ *.env* ចំកណ្តាលពីកូដ។
- **openai**, បណ្ណាល័យនេះគឺជាអ្វីដែលអ្នកនឹងប្រើដើម្បីអន្តរកម្មជាមួយ OpenAI API។
- **pillow**, ដើម្បីធ្វើការជាមួយរូបភាពនៅក្នុងភាសា Python។


1. បង្កើតឯកសារ *.env* ជាមួយមាតិកាដូចខាងក្រោម៖

    ```text
    OPENAI_API_KEY='<add your OpenAI key here>'
    ```



1. ប្រមូលបណ្ណាល័យខាងលើទាំងអស់ក្នុងឯកសារមួយដែលមានឈ្មោះ *requirements.txt* ដូចខាងក្រោម៖

    ```text
    python-dotenv
    openai
    pillow
    ```


1. បន្ទាប់មក បង្កើតបរិស្ថានវេរុឆូអាល់ហើយតំឡើងបណ្ណាល័យទាំងនេះ៖


`````python
# create virtual env
! python3 -m venv venv
# activate environment
! source venv/bin/activate
# install libraries
# pip install -r requirements.txt, if using a requirements.txt file 
! pip install python-dotenv openai pillow
`````

> [!NOTE]
> សម្រាប់ Windows ចុចបញ្ជារប្រមាណខាងក្រោមដើម្បីបង្កើតនិងដំណើរការ virtual environment របស់អ្នក៖

    ```bash
    python3 -m venv venv
    venv\Scripts\activate.bat
    ````

1. បន្ថែមកូដខាងក្រោមទៅក្នុងឯកសារដែលហៅថា *app.py*:

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    import openai

    # នាំចូល dotenv
    dotenv.load_dotenv()

    # បង្កើតវត្ថុ OpenAI (អាន OPENAI_API_KEY ពី .env របស់អ្នក)
    client = openai.OpenAI()


    try:
        # បង្កើតរូបភាពដោយប្រើ API បង្កើតរូបភាព
        generation_response = client.images.generate(
            model="gpt-image-1",
            prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # បញ្ចូលអត្ថបទបញ្ចូលរបស់អ្នកនៅទីនេះ
            size='1024x1024',
            n=1
        )
        # កំណត់ថតសម្រាប់រក្សាទុករូបភាព
        image_dir = os.path.join(os.curdir, 'images')

        # ប្រសិនបើថតមិនមាន សូមបង្កើតវា
        if not os.path.isdir(image_dir):
            os.mkdir(image_dir)

        # ចាប់ផ្ដើមផ្លូវរូបភាព (សូមចំណាំថាប្រភេទឯកសារត្រូវតែ png)
        image_path = os.path.join(image_dir, 'generated-image.png')

        # ម៉ូឌែល gpt-image ខ្ចប់រូបភាពជាផ្នែក base64 (b64_json) មិនមែនជា URL
        image_b64 = generation_response.data[0].b64_json
        generated_image = base64.b64decode(image_b64)
        with open(image_path, "wb") as image_file:
            image_file.write(generated_image)

        # បង្ហាញរូបភាពក្នុងកម្មវិធីមើលរូបភាពលំនាំដើម
        image = Image.open(image_path)
        image.show()

    # ចាប់ករណីកំហុស
    except openai.BadRequestError as err:
        print(err)

    ```

តោះពន្យល់ពីកូដនេះ៖

- ជាលើកដំបូង យើងនាំចូលបណ្ណាល័យដែលយើងត្រូវការជា OpenAI library, dotenv library, ម៉ូឌុល base64, និង Pillow library។

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    import openai
    ```

- បន្ទាប់មក យើងបង្កើត client ដែលអាន API key ពីឯកសារ ``.env`` របស់អ្នក។

    ```python
    # បង្កើតវត្ថុ OpenAI
    client = openai.OpenAI()
    ```

- បន្ទាប់មក យើងបង្កើតរូបភាពនេះ៖

    ```python
    generation_response = client.images.generate(
        model="gpt-image-1",
        prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # បញ្ចូលអត្ថបទបញ្ចូលរបស់អ្នកនៅទីនេះ
        size='1024x1024',
        n=1
    )
    ```

    ម៉ូឌែល `gpt-image` ផ្ដល់រូបភាពជា String **base64** នៅក្នុង `data[0].b64_json`។ យើង decode ទៅជា bytes ហើយសរសេរចូលឯកសារ - គ្មាន URL សម្រាប់ទាញយក។

    ```python
    image_b64 = generation_response.data[0].b64_json
    generated_image = base64.b64decode(image_b64)
    with open(image_path, "wb") as image_file:
        image_file.write(generated_image)
    ```

- ចុងក្រោយ យើងបើករូបភាព ហើយប្រើកម្មវិធីមើលរូបភាពស្តង់ដារដើម្បីបង្ហាញវា៖

    ```python
    image = Image.open(image_path)
    image.show()
    ```

### ព័ត៌មានលម្អិតបន្ថែមអំពីការបង្កើតរូបភាព

តោះមើលប៉ារ៉ាម៉ែត្រ នៃ `images.generate`៖

- **model** ជា ម៉ូឌែលរូបភាពដែលត្រូវប្រើ (ឧទាហរណ៍ `gpt-image-1`)។
- **prompt** ជាអត្ថបទពណ៌នាដែលប្រើបង្កើតរូបភាព។ នៅទីនេះគឺជា "Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils"។
- **size** ជារូបភាពដែលបានបង្កើត (ឧទាហរណ៍ `1024x1024`, `1536x1024`, `1024x1536`, ឬ `"auto"`)។
- **n** ជាចំនួនរូបភាពដែលបានបង្កើត។ នៅទីនេះយើងបង្កើតមួយ។

> ម៉ូឌែលរូបភាពមិនខ្វក់ប៉ារ៉ាម៉ែត្រ `temperature` ទេ - វាជាការគ្រប់គ្រងបង្កើតអត្ថបទ។ ដើម្បីបានការបែបផ្សេងៗ ជាន់ API ម្តងទៀត; ដើម្បីកាត់បន្ថយការបែបផ្សេងៗ បង្កើត prompt ឲ្យជាក់លាក់។

## សមត្ថភាពបន្ថែមនៃការបង្កើតរូបភាព

អ្នកបានឃើញរបៀបបង្កើតរូបភាពជាមួយបន្ទាត់ Python មួយចំនួន។ ម៉ូឌែល `gpt-image` ក៏អាច **កែសម្រួល** រូបភាពដែលមានប្រើប្រាស់។ ដោយផ្តល់រូបភាព, **mask** ជាជម្រើស (ដែលសំគាល់តំបន់ដែលត្រូវកែប្រែ), និង prompt អ្នកអាចបម្លែងផ្នែកមួយរបស់រូបភាព - ឧទាហរណ៍ បន្ថែមមួកទៅក៏បាន។

```python
response = client.images.edit(
  model="gpt-image-1",
  image=open("base_image.png", "rb"),
  mask=open("mask.png", "rb"),
  prompt="An image of a rabbit with a hat on its head.",
  n=1,
  size="1024x1024"
)
# ការកែសម្រួលក៏ត្រូវបានផ្ញើត្រឡប់ជារូបមន្តbase64 ផងដែរ
edited_image = base64.b64decode(response.data[0].b64_json)
```
