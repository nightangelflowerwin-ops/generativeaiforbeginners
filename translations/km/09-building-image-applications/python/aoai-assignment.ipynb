# ការបង្កើតកម្មវិធីបង្កើតរូបភាព (Azure OpenAI)

មានអ្វីៗផ្សេងទៀតសម្រាប់ LLMs មិនមែនគ្រាន់តែបង្កើតអត្ថបទទេ។ អ្នកអាចបង្កើតរូបភាពពីការពិពណ៌នាអត្ថបទផងដែរ។ រូបភាពដែលជារបៀបបង្ហាញ គឺមានប្រយោជន៍នៅក្នុងវិស័យ MedTech, វិស្វកម្ម, ទេសចរណ៍, ការអភិវឌ្ឍល្បែង, ម៉ារកេទីង និងផ្សេងៗទៀត។ ក្នុងមេរៀននេះ យើងបង្កើតជាមួយគំរូ **GPT Image** ថ្មីៗលើ Microsoft Foundry។

## គោលដៅការសិក្សា

នៅចុងបញ្ចប់មេរៀននេះ អ្នកនឹងអាច:

- ធ្វើការពន្យល់ពីអ្វីទៅជាការបង្កើតរូបភាព និងវិញកន្លែងដែលវាប្រើបាន។
- យល់ដឹងពីគំរូ `gpt-image` និងភាពខុសគ្នារវាងគំរូនេះ និងគំរូ DALL·E ចាស់ៗ។
- បង្កើតកម្មវិធីបង្កើតរូបភាព និងកែរូបភាព។

## តើអ្វីទៅជាការបង្កើតរូបភាព?

គំរូបង្កើតរូបភាពបង្កើតរូបភាពមកពីការបញ្ចូលអត្ថបទ។ គំរូសម័យថ្មីដូចជា `gpt-image` សិក្សាខ្សែទំនាក់ទំនងរវាងអត្ថបទ និងរូបភាពនៅពេលបណ្តុះបណ្តាល បន្ទាប់មកវាបំបែកសំឡេងចៃដន្យទៅជារូបភាពដែលផ្គូផ្សព្វគ្នាជាមួយការបញ្ចូលរបស់អ្នក។

គំរូរូបភាពដែលល្បីឈ្មោះពីរប្រភេទគឺ:

- **`gpt-image` (OpenAI)** - ជាទំនើបបច្ចុប្បន្ន ដែលប្រើនៅក្នុងមេរៀននេះ។ វាគាំទ្រការបម្លែងអត្ថបទទៅរូបភាព និងកែសម្រួលរូបភាព (inpainting ជាមួយម៉ាស់)។
- **Midjourney** - គំរូដែលមានប្រជាប្រិយភាពពីភាគីទីបី មានសេវាកម្មផ្ទាល់ និងដំណើរការក្នុង Discord។

> គំរូរូបភាព OpenAI ចាស់ៗ ​- **DALL·E 2** និង **DALL·E 3** - គឺជារបស់ចាស់។ DALL·E 3 មិនអាចប្រើសម្រាប់ការដាក់បញ្ចូលថ្មីៗទេ ហើយមុខងារ `create_variation` មានត្រឹមតែ DALL·E 2 ប៉ុណ្ណោះ។ សូមប្រើគំរូ `gpt-image` សម្រាប់កម្មវិធីថ្មី។

នៅ Microsoft Foundry, **`gpt-image-2`** គឺជាគំរូរូបភាពថ្មី និងមានសមត្ថភាពខ្ពស់បំផុត ហើយត្រូវបានផ្តល់អនុសាសន៍ឱ្យប្រើជាដើម។ `gpt-image-1.5` និង `gpt-image-1-mini` ក៏មានស្រាប់ដោយទូទៅផងដែរ។

> **សំខាន់៖** គំរូ `gpt-image` ឆ្លើយតបទៅជារូបភាពដែលបាននិម្មិតជារាង **base64** (`b64_json`), មិនមែនជาลីង URL ទេ។ កូដរបស់អ្នកកូដអូសខ្សែអត្ថបទ base64 ទៅជា bytes ហើយរក្សាទុក វាមិនមាន URL រូបភាពសម្រាប់ទាញយកទេ។


## ការបង្កើតកម្មវិធីបង្កើតរូបភាពដំបូងរបស់អ្នក

តើត្រូវការអ្វីខ្លះដើម្បីបង្កើតកម្មវិធីបង្កើតរូបភាព? អ្នកត្រូវការបណ្ណាល័យខាងក្រោម៖

- **python-dotenv**, អ្នកត្រូវបានណែនាំយ៉ាងខ្លាំងឲ្យប្រើបណ្ណាល័យនេះ ដើម្បីរក្សាឯកជនភាពរបស់អ្នកនៅក្នុងឯកសារ *.env* ឆ្ងាយពីកូដ។
- **openai**, បណ្ណាល័យនេះជាដើមដែលអ្នកនឹងប្រើដើម្បី​ទំនាក់ទំនងជាមួយ OpenAI API។
- **pillow**, ដើម្បីធ្វើការជាមួយរូបភាពនៅក្នុង Python។

ប្រសិនបើមិនបានធ្វើរួចទេ សូមធ្វើតាមការណែនាំនៅលើទំព័រ [Microsoft Learn](https://learn.microsoft.com/azure/ai-foundry/openai/how-to/create-resource?pivots=web-portal&WT.mc_id=academic-105485-koreyst) ដើម្បីបង្កើតធនធាន Microsoft Foundry និងម៉ូឌែល។ ជ្រើស **gpt-image-2** ជាម៉ូឌែល (ម៉ូឌែលរូបភាព Azure OpenAI ថ្មីបំផុត; DALL·E គឺជាម៉ូឌែលចាស់)។

1. បង្កើតឯកសារ *.env* មានខ្លឹមសារដូចខាងក្រោម៖

    ```text
    AZURE_OPENAI_ENDPOINT=<your endpoint>
    AZURE_OPENAI_API_KEY=<your key>
    AZURE_OPENAI_DEPLOYMENT="gpt-image-2"
    ```

    ស្វែងរកព័ត៌មាននេះនៅក្នុងផ្លូវច្រក Microsoft Foundry សម្រាប់ធនធានរបស់អ្នកនៅក្នុងផ្នែក "Deployments"។



1. ប្រមូលបណ្ណាល័យខាងលើទាំងអស់ឲ្យបានក្នុងឯកសារដែលមានឈ្មោះ *requirements.txt* ដូចខាងក្រោម៖

    ```text
    python-dotenv
    openai
    pillow
    ```


1. បន្ទាប់មក បង្កើតបរិក្ខារស្ម័គ្រចិត្តមួយហើយដំឡើងបណ្ណាល័យទាំងនោះ៖


`````python
# create virtual env
! python3 -m venv venv
# activate environment
! source venv/bin/activate
# install libraries
# pip install -r requirements.txt, if using a requirements.txt file 
! pip install python-dotenv openai pillow
`````

> [!NOTE]
> សម្រាប់ Windows, ប្រើពាក្យបញ្ជាខាងក្រោមដើម្បីបង្កើត និងបង្ហោះបរិស្ថានវីរុសរបស់អ្នក:

    ```bash
    python3 -m venv venv
    venv\Scripts\activate.bat
    ````

1. បន្ថែមកូដខាងក្រោមក្នុងឯកសារដែលហៅថា *app.py*:

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    from openai import AzureOpenAI, BadRequestError

    # នាំចូល dotenv
    dotenv.load_dotenv()

    # កំណត់រចនាសម្ព័ន្ធអតិថិជន Azure OpenAI (Microsoft Foundry)។
    # ម៉ូឌែលរូបភាពត្រូវការកំណែ API ថ្មីៗ - សូមពិនិត្យឯកសារ Foundry សម្រាប់កំណែដែលម៉ូឌែលរបស់អ្នកត្រូវការ។
    client = AzureOpenAI(
      azure_endpoint = os.environ["AZURE_OPENAI_ENDPOINT"],
      api_key=os.environ['AZURE_OPENAI_API_KEY'],
      api_version = "2025-04-01-preview"
      )

    try:
        # បង្កើតរូបភាពដោយប្រើ API ការបង្កើតរូបភាព
        generation_response = client.images.generate(
            prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # បញ្ចូលអត្ថបទបញ្ជារបស់អ្នកនៅទីនេះ
            size='1024x1024',
            n=1,
            model=os.environ['AZURE_OPENAI_DEPLOYMENT']   # ឧ. "gpt-image-2"
        )
        # កំណត់ថតសម្រាប់រក្សាទុករូបភាព
        image_dir = os.path.join(os.curdir, 'images')

        # ប្រសិនបើថតមិនមាន សូមបង្កើតវា
        if not os.path.isdir(image_dir):
            os.mkdir(image_dir)

        # ចាប់ផ្តើមផ្លូវរូបភាព (ចំណាំប្រភេទឯកសារអាចត្រូវជា png)
        image_path = os.path.join(image_dir, 'generated-image.png')

        # ម៉ូឌែល gpt-image បញ្ជូនរូបភាពជាគោលដៅ base64 (b64_json) មិនមែន URL ទេ
        image_b64 = generation_response.data[0].b64_json
        generated_image = base64.b64decode(image_b64)
        with open(image_path, "wb") as image_file:
            image_file.write(generated_image)

        # បង្ហាញរូបភាពនៅក្នុងកម្មវិធីមើលរូបភាពលំនាំដើម
        image = Image.open(image_path)
        image.show()

    # ចាប់ករណីខុសប្លែក
    except BadRequestError as err:
        print(err)

    ```

មកពន្យល់កូដនេះ៖

- ជាដំបូង យើងនាំចូលបណ្ណាល័យដែលយើងត្រូវការព្រមទាំងបណ្ណាល័យ OpenAI, បណ្ណាល័យ dotenv, ម៉ូឌុល base64, និងបណ្ណាល័យ Pillow។

    ```python
    import os
    import base64
    from PIL import Image
    import dotenv
    from openai import AzureOpenAI, BadRequestError
    ```

- បន្ទាប់មក, យើងដំណើរការរឿងបរិភោគបរិស្ថានពីឯកសារ *.env*។

    ```python
    # នាំចូល dotenv
    dotenv.load_dotenv()
    ```

- បន្ទាប់ពីនេះ, យើងកំណត់លំនាំឲ្យអតិថិជន Azure OpenAI (Microsoft Foundry)។

    ```python
    client = AzureOpenAI(
      azure_endpoint = os.environ["AZURE_OPENAI_ENDPOINT"],
      api_key=os.environ['AZURE_OPENAI_API_KEY'],
      api_version = "2025-04-01-preview"
      )
    ```

- បន្ទាប់មក យើងបង្កើតរូបភាព៖

    ```python
    generation_response = client.images.generate(
        prompt='Bunny on horse, holding a lollipop, on a foggy meadow where it grows daffodils',    # បញ្ចូលអត្ថបទ​បញ្ចូលរបស់អ្នកនៅទីនេះ
        size='1024x1024',
        n=1,
        model=os.environ['AZURE_OPENAI_DEPLOYMENT']
    )
    ```

    ម៉ូឌែល `gpt-image` ធ្វើចេញរូបភាពជាខ្សែអក្សរ **base64** នៅក្នុង `data[0].b64_json`។ យើងបំលែងវាទៅជាបៃ និងសរសេរចូលឯកសារ - គ្មាន URL សម្រាប់ទាញយកឡើយ។

    ```python
    image_b64 = generation_response.data[0].b64_json
    generated_image = base64.b64decode(image_b64)
    with open(image_path, "wb") as image_file:
        image_file.write(generated_image)
    ```

- ចុងក្រោយ យើងបើករូបភាព និងប្រើកម្មវិធីមើលរូបភាពស្តង់ដារដើម្បីបង្ហាញវា៖

    ```python
    image = Image.open(image_path)
    image.show()
    ```

### ព័ត៌មានលម្អិតបន្ថែមអំពីការបង្កើតរូបភាព

មកមើលប៉ារ៉ាម៉ែត្រនៃ `images.generate`៖

- **prompt**, គឺជាអត្ថបទនៃការបញ្ចូលសម្រាប់បង្កើតរូបភាព។ នៅទីនេះវាគឺ "ក្រពើលើសេះ កាន់ឈុតក្រឡុក នៅចំលើយពពក ដែលមានផ្កា daffodils ជាច្រើន"។
- **size**, គឺជា ទំហំរូបភាពដែលបានបង្កើត (ឧទាហរណ៍ `1024x1024`, `1536x1024`, `1024x1536`, ឬ `"auto"` )។
- **n**, គឺជាចំនួនរូបភាពដែលបានបង្កើត។ នៅទីនេះយើងបង្កើតមួយរូប។
- **model**, គឺជាឈ្មោះការតំឡើងម៉ូឌែលរូបភាពរបស់អ្នក (ឧទាហរណ៍ `gpt-image-2`)។

> ម៉ូឌែលរូបភាពមិនមានប៉ារ៉ាម៉ែត្រ `temperature` ទេ - វាជាការត្រួតពិនិត្យសម្រាប់ការបង្កើតអត្ថបទ។ ដើម្បីទទួលបានភាពចម្រុះ, ហៅ API ម្តងទៀត; ដើម្បីកាត់បន្ថយភាពចម្រុះ, ធ្វើឲ្យ prompt របស់អ្នកមានលម្អិតច្បាស់។

## សមត្ថភាពបន្ថែមនៃការបង្កើតរូបភាព

អ្នកបានឃើញរបៀបបង្កើតរូបភាពជាមួយខ្សែ Python ចំនួនប៉ុន្មាន។ ម៉ូឌែល `gpt-image` ក៏អាច **កែសម្រួល** រូបភាពមាន រួចហើយបានផងដែរ។ ដោយផ្ដល់រូបភាព មួយ **mask** ជាជម្រើស (ដែលសម្គាល់តំបន់ដែលចង់ផ្លាស់ប្ដូរ), និង prompt មួយ អ្នកអាចផ្លាស់ប្ដូរផ្នែកមួយនៃរូបភាព - ឧទាហរណ៍ បង្កើតមួកដាក់លើក្រពើរបស់យើង។

```python
response = client.images.edit(
  model=os.environ['AZURE_OPENAI_DEPLOYMENT'],
  image=open("base_image.png", "rb"),
  mask=open("mask.png", "rb"),
  prompt="An image of a rabbit with a hat on its head.",
  n=1,
  size="1024x1024"
)
# ការកាត់តមួយចំនួនត្រូវបានត្រឡប់មកវិញជា base64 ផងដែរ
edited_image = base64.b64decode(response.data[0].b64_json)
```
